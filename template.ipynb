```python
import pandas as pd
import numpy as np
from pathlib import Path

import plotly.express as px
pd.options.plotting.backend = 'plotly'

from dsc80_utils import * # Feel free to uncomment and use this.
```

```python
tracks = pd.read_csv("music_tracks.csv")
artists = pd.read_csv("artists.csv")
```

```python
tracks = tracks.drop(columns=[
    "track_id",
    "duration_ms",
    "explicit",
    "loudness",
    "mode",
    "acousticness",
    "instrumentalness",
    "tempo",
    "time_signature", 
])
tracks = tracks.drop(columns=["Unnamed: 0"])

tracks = tracks.dropna()
```

```python
artists = artists.drop(columns= 'id')
artists = artists.dropna() 
```

```python
merged = tracks.merge(
    artists,
    left_on="artists",
    right_on="name",
    how="left"
)
merged_clean = merged.dropna(subset=["name"])

merged_clean = merged_clean.rename(columns={
    "popularity_x": "track_popularity",
    "popularity_y": "artist_popularity"
})

merged_clean
```

```python
genre_values = (
    merged_clean["genres"]
    .astype(str)
    .str.replace("[", "", regex=False)
    .str.replace("]", "", regex=False)
    .str.replace("'", "", regex=False)
    .str.split(", ")
)

all_genres = []

for genre_list in genre_values:
    for genre in genre_list:
        all_genres.append(genre)

genre_counts = (
    pd.DataFrame({"genre": all_genres})
    .groupby("genre")
    .size()
    .reset_index(name="count")
    .sort_values("count", ascending=False)
)

hiphop_genre_counts = genre_counts[
    genre_counts["genre"].str.contains("hip hop|hip-hop|rap", case=False, na=False)
]

us_hiphop_genres = [
    "pop rap",
    "sad rap",
    "emo rap",
    "rap",
    "hip hop",
    "gangster rap",
    "hardcore hip hop",
    "east coast hip hop",
    "miami hip hop",
    "deep underground hip hop",
    "alternative hip hop",
    "southern hip hop",
    "trap",
    "melodic rap",
    "dark trap",
    "kentucky hip hop",
    "conscious hip hop",
    "meme rap",
    "comedy rap",
    "underground hip hop",
    "lgbtq+ hip hop",
    "minnesota hip hop",
    "industrial hip hop",
    "queens hip hop",
    "experimental hip hop",
    "old school hip hop",
    "country rap",
    "chicago rap",
    "florida rap",
    "philly rap",
    "new jersey rap",
    "pittsburgh rap",
    "christian hip hop",
    "nyc rap",
    "christian trap",
    "atl hip hop",
    "detroit hip hop",
    "west coast trap",
    "political hip hop",
    "vapor trap",
    "baton rouge rap",
    "jazz rap",
    "indie hip hop",
    "cincinnati rap",
    "harlem hip hop",
    "viral rap",
    "asian american hip hop",
    "portland hip hop",
    "dirty south rap",
    "birmingham hip hop",
    "ohio hip hop",
    "psychedelic hip hop",
    "north carolina hip hop",
    "bronx hip hop",
    "tennessee hip hop",
    "deep southern trap",
    "new orleans rap",
    "oakland hip hop",
    "sacramento hip hop",
    "iowa hip hop",
    "san antonio rap",
    "military rap",
    "flint hip hop",
    "rhode island rap",
    "dmv rap"
]
```

```python
def has_us_hiphop_genre(genre_string):
    genre_string = str(genre_string)
    
    for genre in us_hiphop_genres:
        if f"'{genre}'" in genre_string:
            return True
    
    return False

hiphop_artists = merged_clean[
    merged_clean["genres"].apply(has_us_hiphop_genre)
]

hiphop_artists = hiphop_artists.assign(
    song_type=np.where(
        hiphop_artists["track_genre"] == "hip-hop",
        "Hip-hop songs",
        "Non-hip-hop songs"
    )
)
```

```python
popularity_data = hiphop_artists[["track_popularity", "song_type"]].dropna()

popularity_data.groupby("song_type")["track_popularity"].agg(["count", "mean"])
```

```python
def difference_in_means(data):
    means = data.groupby("song_type")["track_popularity"].mean()
    return means["Non-hip-hop songs"] - means["Hip-hop songs"]

observed_stat = difference_in_means(popularity_data)
observed_stat
```

```python
n_repetitions = 1000
simulated_stats = []

for i in range(n_repetitions):
    shuffled_labels = np.random.permutation(popularity_data["song_type"])
    
    shuffled = popularity_data.assign(
        shuffled_song_type=shuffled_labels
    )
    
    means = shuffled.groupby("shuffled_song_type")["track_popularity"].mean()
    stat = means["Non-hip-hop songs"] - means["Hip-hop songs"]
    
    simulated_stats.append(stat)

simulated_stats = np.array(simulated_stats)
```

```python
p_value = np.mean(simulated_stats >= observed_stat)
p_value
```

```python
plt.hist(simulated_stats, bins=30)
plt.axvline(observed_stat)
plt.xlabel("Difference in Mean Popularity")
plt.ylabel("Frequency")
plt.title("Permutation Test: Non-Hip-Hop Popularity - Hip-Hop Popularity")
plt.show()
```
